```python
#Q1: KIỂM TRA SỐ NGUYÊN TỐ
import math

def kiem_tra_so_nguyen(n):
  if n <= 1:
    return False
  for i in range (2, int(math.isqrt(n)) + 1):
    if n % 2 == 0:
      return False

    return True


n = int(input('Mời nhập số:'))

if kiem_tra_so_nguyen(n):
  print(f'Nó là 1 số nguyên tố: {n}')
else:
  print(f'Nó không phải là 1 số nguyên tố: {n}')
```

```python
#Q2: hàm tinh_giai_thua(n)

def giai_thua(n):
    result = 1;
    for i in range (1, n + 1):
        result *= i

    return result

n = int(input("Nhập số nguyên dương n: "))

if n < 0:
    print("Vui lòng nhập số nguyên không âm.")
else:
    print(f"{n}! = {giai_thua(n)}")


```

```python
#Q3: Tình tổng các chữ số

def tinh_tong(n):
    sum = 0

    for ky_tu in str(n):
        sum += int(ky_tu)
    return sum

# Hoặc viết gọn 1 dòng:
# def tinh_tong(n):
#     return sum(int(ch) for ch in str(n))

n = int(input('Mời nhập số nguyên dương: '))

if n <= 0:
    print('Đây không phải là số nguyên dương')
else:
    print(f'Tổng các chữ số của {n} là: {tinh_tong(n)}')
```

```python
#Q4: Viết một hàm tinh_trung_binh(data)

def tinh_trung_binh(data):

  if not data:
    return 0

  sum = 0
  for s in data:
    sum += s
  return sum / len(data)

danh_sach = [10, 20, 30, 40, 50]
ket_qua = tinh_trung_binh(danh_sach)

print(f"Danh sách: {danh_sach}")
print(f"Giá trị trung bình cộng: {ket_qua}")


```

```python
#Q5: dem_so_lan_xuat_hien(data)
def dem_so_lan_xuat_hien(data):
    count = {}

    for num in data:
        if num in count:
           count[num] += 1
        else:
          count[num] = 1

  # In kết quả ra màn hình
    for num, so_lan in count.items():
        print(f"Số {num} xuất hiện {so_lan} lần")

# --- Chạy thử ---
data = [1, 1, 2, 3, 2, 4, 5, 5, 5, 5]
dem_so_lan_xuat_hien(data)
```
